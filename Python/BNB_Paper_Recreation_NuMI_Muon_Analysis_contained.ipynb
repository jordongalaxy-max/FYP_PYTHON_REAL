```python
# Fit NuMI reconstructed muon momentum with several candidate distributions
from scipy import stats

# Use positive values only for continuous positive-support distributions
x = muon_p["reco_muon_p"].dropna()
x = x[x > 0].to_numpy()

# Candidate distributions: shape-flexible and commonly used for momentum-like spectra
candidates = {
    "normal": stats.norm,
    "lognorm": stats.lognorm,
    "gamma": stats.gamma,
    "weibull_min": stats.weibull_min,
}

fit_rows = []
for name, dist in candidates.items():
    params = dist.fit(x)

    # Log-likelihood and AIC
    logpdf = dist.logpdf(x, *params)
    logpdf = np.where(np.isfinite(logpdf), logpdf, -1e12)
    logL = np.sum(logpdf)
    k = len(params)
    aic = 2 * k - 2 * logL

    # KS goodness-of-fit test
    ks_stat, ks_p = stats.kstest(x, dist.cdf, args=params)

    fit_rows.append(
        {
            "distribution": name,
            "AIC": aic,
            "KS_stat": ks_stat,
            "KS_pvalue": ks_p,
            "params": params,
        }
    )

fit_table = pandas.DataFrame(fit_rows).sort_values(["KS_stat", "AIC"], ascending=[True, True]).reset_index(drop=True)
print("Candidate fit ranking (best on top):")
print(fit_table[["distribution", "AIC", "KS_stat", "KS_pvalue"]].to_string(index=False))

best_name = fit_table.loc[0, "distribution"]
best_params = fit_table.loc[0, "params"]
best_dist = candidates[best_name]
print(f"\nBest fit: {best_name}")
print(f"Best-fit parameters: {best_params}")

# Plot normalized histogram + fitted PDF
x_plot = np.linspace(0, 6, 600)
pdf_plot = best_dist.pdf(x_plot, *best_params)

plt.figure(figsize=(10, 6))
plt.hist(x, bins=50, range=(0, 6), density=True, alpha=0.5, color="steelblue", label="NuMI reco muon momentum")
plt.plot(x_plot, pdf_plot, "r-", lw=2.0, label=f"Best fit: {best_name}")
plt.xlabel("Muon Momentum (GeV/c)")
plt.ylabel("Normalized density")
plt.title("NuMI Reco Muon Momentum: Best-Fit Distribution")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
# Optimized and stable fit: bounded 2-component log-normal mixture
from scipy import stats, optimize

x = muon_p["reco_muon_p"].dropna()
x = x[x > 0].to_numpy()
logx = np.log(x)

# Baseline: single log-normal (loc fixed at 0 for x>0)
s1, loc1, scale1 = stats.lognorm.fit(x, floc=0)
logL_single = np.sum(stats.lognorm.logpdf(x, s1, loc=loc1, scale=scale1))
aic_single = 2 * 2 - 2 * logL_single


def mixture_pdf(xv, w, mu1, sig1, mu2, sig2):
    c1 = stats.lognorm.pdf(xv, s=sig1, loc=0, scale=np.exp(mu1))
    c2 = stats.lognorm.pdf(xv, s=sig2, loc=0, scale=np.exp(mu2))
    return w * c1 + (1 - w) * c2


def mixture_cdf(xv, w, mu1, sig1, mu2, sig2):
    c1 = stats.lognorm.cdf(xv, s=sig1, loc=0, scale=np.exp(mu1))
    c2 = stats.lognorm.cdf(xv, s=sig2, loc=0, scale=np.exp(mu2))
    return w * c1 + (1 - w) * c2


# Direct bounded optimization to prevent degenerate components
# theta = [w, mu1, sigma1, mu2, sigma2]
def nll(theta):
    w, mu1, sig1, mu2, sig2 = theta
    pdf = mixture_pdf(x, w, mu1, sig1, mu2, sig2)
    pdf = np.clip(pdf, 1e-300, None)
    return -np.sum(np.log(pdf))


# Bounds chosen to be physically reasonable for NuMI muon momentum in [0, 6] GeV/c
# log-scale centers around log(0.05) to log(8) and widths kept finite
bounds = [
    (0.02, 0.98),   # mixture weight
    (-3.0, 2.2),    # mu1
    (0.12, 1.4),    # sigma1
    (-3.0, 2.2),    # mu2
    (0.12, 1.4),    # sigma2
]

rng = np.random.default_rng(123)
starts = []
for _ in range(16):
    starts.append([
        rng.uniform(0.2, 0.8),
        np.quantile(logx, rng.uniform(0.10, 0.45)),
        rng.uniform(0.25, 0.9),
        np.quantile(logx, rng.uniform(0.55, 0.95)),
        rng.uniform(0.25, 0.9),
    ])

best = None
for theta0 in starts:
    res = optimize.minimize(nll, theta0, method="L-BFGS-B", bounds=bounds)
    if not res.success:
        continue
    if best is None or res.fun < best.fun:
        best = res

if best is None:
    raise RuntimeError("Bounded mixture optimization failed for all starts.")

w, mu1, sig1, mu2, sig2 = best.x

# Sort components by scale for readability
if mu1 > mu2:
    w, mu1, sig1, mu2, sig2 = (1 - w), mu2, sig2, mu1, sig1

logL_mix = -best.fun
aic_mix = 2 * 5 - 2 * logL_mix

ks_single = stats.kstest(x, lambda t: stats.lognorm.cdf(t, s1, loc=0, scale=scale1))
ks_mix = stats.kstest(x, lambda t: mixture_cdf(t, w, mu1, sig1, mu2, sig2))

summary = pandas.DataFrame(
    [
        {
            "model": "1x lognorm",
            "AIC": aic_single,
            "KS_stat": ks_single.statistic,
            "KS_pvalue": ks_single.pvalue,
        },
        {
            "model": "2x lognorm mixture (bounded)",
            "AIC": aic_mix,
            "KS_stat": ks_mix.statistic,
            "KS_pvalue": ks_mix.pvalue,
        },
    ]
)

print("Fit comparison (smaller KS_stat and AIC are better):")
print(summary.to_string(index=False))

print("\nStable mixture parameters:")
print(f"w={w:.4f}, mu1={mu1:.4f}, sigma1={sig1:.4f}, mu2={mu2:.4f}, sigma2={sig2:.4f}")
print(f"component scales: exp(mu1)={np.exp(mu1):.4f}, exp(mu2)={np.exp(mu2):.4f}")

x_plot = np.linspace(0.001, 6, 800)
pdf_single = stats.lognorm.pdf(x_plot, s1, loc=0, scale=scale1)
pdf_mix = mixture_pdf(x_plot, w, mu1, sig1, mu2, sig2)

plt.figure(figsize=(10, 6))
plt.hist(x, bins=50, range=(0, 6), density=True, alpha=0.45, color="steelblue", label="NuMI reco muon momentum")
plt.plot(x_plot, pdf_single, color="orange", lw=2, label="Single lognorm")
plt.plot(x_plot, pdf_mix, color="crimson", lw=2.2, label="Optimized bounded 2x lognorm")
plt.xlabel("Muon Momentum (GeV/c)")
plt.ylabel("Normalized density")
plt.title("NuMI Muon Momentum: Optimized Stable Fit")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
#BNB, input000.flat.root
#yy-shower truth selection 

#Importing relevant libraries
import numpy as np
import uproot
import pandas
import hist
import matplotlib.pyplot as plt

#enable this to see all outputs in a cell
#from IPython.core.interactiveshell import InteractiveShell
#InteractiveShell.ast_node_interactivity = "all"

#Opening the relevant ROOT file and tree
file = uproot.open(
    "e:/FYP/medulla_outputs/2_Shower_Selection_flashcut/ccpi0ana_icarus.root:events/cv/selected_nu"
)

#showing all TBranches
file.keys()

#Purity analysis
# --------------------------------------------------------------------------------

#Combining interaction category variable and pi0 mass variable for easier analysis
combinedcategory = file.arrays(["true_category_topology_ccpi0_simple2","reco_pi0_mass"],library="pd")

#Removing interactions with pi0 mass greater than 400 MeV/c^2 as they are not relevant for our analysis and are likely background interactions
combinedcategory = combinedcategory[combinedcategory["reco_pi0_mass"] < 400]
POT = 1.74e20

#Splitting the interaction category variable into the relevant categories

#cosmics category: interactions that are not neutrino interactions (true_category_topology_ccpi0_simple2 is NaN)
cosmics_category = combinedcategory[combinedcategory["true_category_topology_ccpi0_simple2"].isna()]
print(f"Number of cosmic interactions and percentage: {len(cosmics_category)} out of {len(combinedcategory)} ({len(cosmics_category)/len(combinedcategory)*100:.2f}%)")

#neutrino 1mu1pi0 category: Neutrino interactions with 1 primary muon, 0 primary charged pions, and 1 primary neutral pion (true_category_topology_ccpi0_simple2 == 0)
neutrino_1mu0pi1pi0_category = combinedcategory[combinedcategory["true_category_topology_ccpi0_simple2"] == 0]
print(f"Number of 1mu0pi1pi0 CC neutrinointeractions and percentage: {len(neutrino_1mu0pi1pi0_category)} out of {len(combinedcategory)} ({len(neutrino_1mu0pi1pi0_category)/len(combinedcategory)*100:.2f}%)")

#neutrino 1mu0pi2pluspi0 category: Neutrino interactions with 1 primary muon, 0 primary charged pions, and 2 or more primary neutral pions (true_category_topology_ccpi0_simple2 == 1)
neutrino_1mu0pi2pluspi0_category = combinedcategory[combinedcategory["true_category_topology_ccpi0_simple2"] == 1]
print(f"Number of 1mu0pi2+pi0 CC neutrino interactions and percentage: {len(neutrino_1mu0pi2pluspi0_category)} out of {len(combinedcategory)} ({len(neutrino_1mu0pi2pluspi0_category)/len(combinedcategory)*100:.2f}%)")

#neutrino 1muNpiXpi0 category: Neutrino interactions with 1 primary muon, N primary charged pions (N>=1), and any number of primary neutral pions (true_category_topology_ccpi0_simple2 == 2)
neutrino_1muNpiXpi0_category = combinedcategory[combinedcategory["true_category_topology_ccpi0_simple2"] == 2]
print(f"Number of 1muNpiXpi0 CC neutrino interactions and percentage: {len(neutrino_1muNpiXpi0_category)} out of {len(combinedcategory)} ({len(neutrino_1muNpiXpi0_category)/len(combinedcategory)*100:.2f}%)")

#neutrino 0muNpi0 category: Neutrino interactions with 0 primary muons, N primary charged pions (N>=0), and any number of primary neutral pions (true_category_topology_ccpi0_simple2 == 3)
neutrino_0muNpi0_category = combinedcategory[combinedcategory["true_category_topology_ccpi0_simple2"] == 3]
print(f"Number of 0muNpi0 NC neutrino interactions and percentage: {len(neutrino_0muNpi0_category)} out of {len(combinedcategory)} ({len(neutrino_0muNpi0_category)/len(combinedcategory)*100:.2f}%)")

#neutrino othernu category: Neutrino interactions that do not fall into the above categories (true_category_topology_ccpi0_simple2 == 4)
neutrino_othernu_category = combinedcategory[combinedcategory["true_category_topology_ccpi0_simple2"] == 4]
print(f"Number of other neutrino interactions and percentage: {len(neutrino_othernu_category)} out of {len(combinedcategory)} ({len(neutrino_othernu_category)/len(combinedcategory)*100:.2f}%)")

#Plotting the pi0 mass distribution for each category into one histogram
plt.figure(figsize=(10,6))
plt.hist(neutrino_1mu0pi1pi0_category["reco_pi0_mass"], bins=50, range=(0, 400), alpha=0.5, weights=np.ones(len(neutrino_1mu0pi1pi0_category)) / len(neutrino_1mu0pi1pi0_category), label=f"1u0pi1pi0 ({len(neutrino_1mu0pi1pi0_category)/len(combinedcategory)*100:.2f}%)")
plt.hist(neutrino_1mu0pi2pluspi0_category["reco_pi0_mass"], bins=50, range=(0, 400), alpha=0.5, weights=np.ones(len(neutrino_1mu0pi2pluspi0_category)) / len(neutrino_1mu0pi2pluspi0_category), label=f"1u0pi2+pi0 ({len(neutrino_1mu0pi2pluspi0_category)/len(combinedcategory)*100:.2f}%)")
plt.hist(neutrino_1muNpiXpi0_category["reco_pi0_mass"], bins=50, range=(0, 400), alpha=0.5, weights=np.ones(len(neutrino_1muNpiXpi0_category)) / len(neutrino_1muNpiXpi0_category), label=f"1uNpiXpi0 ({len(neutrino_1muNpiXpi0_category)/len(combinedcategory)*100:.2f}%)")
plt.hist(neutrino_0muNpi0_category["reco_pi0_mass"], bins=50, range=(0, 400), alpha=0.5, weights=np.ones(len(neutrino_0muNpi0_category)) / len(neutrino_0muNpi0_category), label=f"NC Npi0 ({len(neutrino_0muNpi0_category)/len(combinedcategory)*100:.2f}%)")
plt.hist(neutrino_othernu_category["reco_pi0_mass"], bins=50, range=(0, 400), alpha=0.5, weights=np.ones(len(neutrino_othernu_category)) / len(neutrino_othernu_category), label=f"Other v ({len(neutrino_othernu_category)/len(combinedcategory)*100:.2f}%)")
plt.hist(cosmics_category["reco_pi0_mass"], bins=50, range=(0, 400), alpha=0.5, weights=np.ones(len(cosmics_category)) / len(cosmics_category), label=f"Cosmics ({len(cosmics_category)/len(combinedcategory)*100:.2f}%)")
plt.legend()
plt.xlabel("Reconstructed pi₀ mass (MeV/c²)")
plt.ylabel("Normalized candidates")
plt.title(f"Reconstructed pi0 mass distribution for different interaction categories ({POT} POT)")
```
